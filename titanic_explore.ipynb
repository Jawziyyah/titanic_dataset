#### Basic Operations

```python
#Simple Filtering

df[df['Fare']>250]
```

```python
#Combined Multiple Conditions
df[(df['Age']>30) & (df['Fare']>250)]
```

```python
#String Condition
df[df['Name'].str.contains("Charles")]
```

```python
#Select all where Sex=Female and Survived=0
df[(df['Sex']=='Male') & (df['Survived']==1)]
```

```python
df[(df['Sex'] == 'Female') & (df['Survived'] == 0)]
```

Group By.


```python
# Group by class
class_group = df.groupby('Pclass')
# This alone is not useful, unless you use operations like .sum, .mean
class_group
```

```python
class_group['Fare'].mean()
```

```python
#Grouping by gender
gender_group=df.groupby('Sex')
gender_group['Survived'].sum()
```

```python
#Grouping by survival
survival_group=df.groupby('Survived')
survival_group['Survived'].sum()
```

```python
#Total fare per class
class_group['Fare'].sum()
```

```python
#Multiple Aggregration
class_group.agg({"Fare":['mean','min','max']})
```

```python
# find the passengers with the max and min fare from each class
class_group.agg({'Fare':['max','min']})
```
